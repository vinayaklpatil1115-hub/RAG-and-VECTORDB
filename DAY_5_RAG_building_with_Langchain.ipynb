```python
# importing libraries
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from google.colab import files, userdata
import os
```

```python
os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY')
```

```python
# pdf loading
uploaded =files.upload()
pdf_name = list(uploaded.keys())[0]
print(pdf_name)
```

```python
loader = PyPDFLoader(pdf_name)
documents = loader.load()
```

```python
print(len(documents))
```

```python
print(documents[0].page_content)
```

```python
print(documents[0].metadata)
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50
)
```

```python
chunks = splitter.split_documents(documents)
```

```python
print(len(chunks))
```

```python
print(" INFO about CHUNK - 5")
print("page content - ")
print(chunks[4].page_content)
print("metadata - ")
print(chunks[4].metadata)
```

```python
import chromadb
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
from langchain_chroma import Chroma
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model = "models/gemini-embedding-001"
)
```

```python
client = chromadb.Client()
```

```python
collection_name = "student_handbook"
```

```python
vectorstore = Chroma.from_documents(
    documents = chunks,
    embedding = embeddings,
    client = client,
    collection_name = collection_name
)
```

```python
# similarity search
```

```python
query = input("Enter Query :")
```

```python
results = vectorstore.similarity_search(
    query = query,
    k = 3
)
```

```python
for i, doc in enumerate(results):
  print(f"{i+1} : {doc.page_content}")
  # print metadata
  print("metadata :",doc.metadata)
  print("-"*50)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
retriever =vectorstore.as_retriever(
    search_kwargs={"k" :3}
    )
```

```python
question = input("Enter Question :")
```

```python
retrieved_doc = retriever.invoke(question)
```

```python
for i, doc in enumerate(retrieved_doc):
  print(f"{i+1} : {doc.page_content}")
  # print metadata
  print("metadata :",doc.metadata)
  print("-"*50)
```

```python
context = "\n".join([doc.page_content for doc in retrieved_doc])
```

```python
print(context)
```

```python
# initialize the llm
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash-lite",
    temperature = 0.2
)
```

```python
prompt = PromptTemplate.from_template("""
        You are a helpful student assistant bot.
        Use the following pieces of context to answer the question at the end.
        If you don't know the answer, just say Sorry! i am unable to solve your query.
        dont make up the answer.

        context : {context}
        question :{question}
        answer :

        """)
```

```python
formatted_prompt = prompt.invoke({
    "context" : context,
    "question" : question
})
```

```python
response =llm.invoke(formatted_prompt)
```

```python
print(response)
```

```python
if isinstance(response.content, list):
  for block in response.content:
    if block.get("type") == "text":
      print(block["text"])
else:

    print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_doc(doc):
  return "\n\n".join([doc.page_content for doc in doc])
```

```python
rag_prompt =PromptTemplate.from_template("""
        You are a helpful student handbook assistant.

Answer the user's question using ONLY the
information provided in the context.

Important instructions:
1. Do not use outside knowledge.
2. Do not assume or invent information.
3. Answer only what the context supports.
4. If the answer is not available in the context,
   say that the information is not available
   in the uploaded document.
5. Make sure your answer is relevant to the
   user's actual question.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
rag_chain = (
    rag_prompt
    | llm
    | StrOutputParser()
)
```

```python
def ask_question(question):
  retrieved_doc = retriever.invoke(question)
  formatted_doc = format_doc(retrieved_doc)
  answer = rag_chain.invoke({
      "context" : formatted_doc,
      "question" : question
  })

  print("chatbot :",answer)
```

```python
while True:
  question = input("Enter Question :")
  if question.lower() == "exit":
    break

  ask_question(question)
```
