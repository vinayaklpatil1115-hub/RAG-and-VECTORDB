```python
documents = [
    "Artificial Intelligence is transforming the healthcare industry.",
    "The Earth revolves around the Sun once every year.",
    "Python is a popular programming language used for machine learning.",
    "Regular exercise helps maintain physical and mental health.",
    "The stock market involves buying and selling company shares.",
    "Solar panels convert sunlight into electrical energy.",
    "Cricket is one of the most popular sports in India.",
    "Online shopping allows customers to purchase products from home.",
    "Photosynthesis is the process by which plants prepare their food.",
    "Cybersecurity protects computer systems from digital attacks."
]
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
doc_embedding = model.encode(documents)
```

```python
doc_embedding.shape
```

```python
query = input("Enter your query: ")
```

```python
query_embedding = model.encode([query])
```

```python
query_embedding.shape
```

```python
similarity = cosine_similarity(
    query_embedding,
    doc_embedding
)[0]
```

```python
best_index = np.argmax(similarity)
```

```python
print("Query : ", query)
print("Best Match : ", documents[best_index])
print("Best match score : ",
      round(similarity[best_index],2))
```
