```python
from sentence_transformers import SentenceTransformer
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
input1 = input("Enter first sentence : ")
input2 = input("Enter second sentence : ")
```

```python
embedding1 = model.encode([input1])
embedding2 = model.encode([input2])
```

```python
print("dim of embedding 1 :",embedding1.shape)
print("dim of embedding 2 :",embedding2.shape)
```

```python
similarity = cosine_similarity(
    embedding1,
    embedding2
)
```

```python
print("sentence 1 :",input1)
print("sentence 2 :",input2)
print("similarity :",
      similarity[[0][0]]
      )
```
